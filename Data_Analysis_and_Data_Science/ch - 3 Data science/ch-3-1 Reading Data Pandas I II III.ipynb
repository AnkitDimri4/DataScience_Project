## Creating copy of Orignal data
There are two ways to create copies
- Shallow copy 
    - Function 
        - samp=cars_data.copy(deep=False),- samp = cars_data
- Description 
    - It only creates a new variables that shares the reference of the orignal object
    - Any changes made to a copy of object will be reflected in the orignal object as well
- Deep copy
    - cars_data1=cars_data.copy(deep=True)
    - in case of deep copy, a copy of object is copied in other object with no refrence to the orignal
    - Any changes  made to copy of object will not be reflected in the orignal object.

### Indexing and selecting data
- Python slicing operator '[]' and attribute/dot operator '.' are used for indexing 
- provide quick and easy access to panda data structures
- The Function head returns the first n rows from the dataframe
    - [DataFrame.head([n])]
- The function tail returns the last n rows for the object based on position
    - [DataFrame.tail(n)]
- To access a scalar value, the fastest way is to use the at and iat methods
    - at provides label-based scalar lookups
        - DataType.at[4,'FuelType']
    -iat provides integer-based lookups
        - DataType.iat[5,6]
- To access a group of rows and columns by label(s).loc[] can be used
    - DataType.loc[:,'FuelType'] [: for the all]

# Pandas Dataframes II
 

- Data Types 
        - The way information gets stored in a dataframe or a python object affects the analysis and outputs of calculation. 
        - Strings are known as objects in pasnds which can store values that contain numbers and / or characters 
    - for example : 'category1'
    - includes integeers and floats
    
    - Numeric Types
        - Pandas and base python uses different names for data types
        - int --- int64  numeric characters
        - float -- float64 -- Numeric characters with decimals
        - '64' simply refers to the memory allocated to store data in each cell which effectively relates to how many digits it can store in each "cell" [64 bits is equivalent to 8 bytes] 
        - Allocating spaces ahead of tims allows computers to optimize storage and processing efficiency.
    
    - character Types
        - Difference b/w category and object
            
            - Category 
                - A string variable consisting of only a few different values. Converting such a string variable to a categorical variable will save some memory 
                - A categorical variable takes on a limited, fixed number of possible values
            
            - Object 
                - The column will be assigned as object data type when it has mixed types (numbers and strings). if a column contains 'nan' (blank cells), pandas will default to object datatypes..
                - For strings, the length is not fixed
    
- Checking data types of each column
    - dtypes returns a series with the data type of each column
        - Syntax : DataFrame.dtypes

- Count of unique data types 
    - get_dtype_counts() returns counts of unique data types in the dataframe
        - Syntax: DataFrame.get_dtype_counts()

- selected data based on data types 
    - pandas.DataFrame.select_dtypes() returns a subset of the cooliumns from dataframe based on the column dtypes
        - Syntax: DataFrame.select_dtypes(include=None, exclude=None)
        
- Concise summary of dataframe
    - info() returns a concise summary of a dataframe

        - data types of columns

        - memory usage
            Syntax: DataFrame.info()
