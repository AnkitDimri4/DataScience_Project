> ### **Introduction to NumPy**
 - NumPy is a Python package and it stands for numerical python
 - Fundamental package for numerical computations in Python
 - Suppose N-dimensional array objects that can be used for processing multidimensional data
 - Supports different data types 
 

#### **Array** 
- An array is a data structure that **stores values of same data type**
- Lists can contain values corresponding to different data types.
- Arrays in Python can only contain values corresponding to same data type

### **NumPy Array** 
- A numpy array is a grid of  values, all of the same type, and is indexed by a tuple of non-negative integers
- The **number of dimension is the rank of the array**
- The shape of an array is a tuple of integers giving the size of the array along each dimension


### **Creation of array**

```python
my_list = [1,2,4,5,6,7]
print(my_list)
```

#### **To create numpy array, we first need to import the numpy package:**

```python
import numpy as np
array = np.array(my_list,dtype = int)
print(array)
```

```python
print(type(array)) ## data type of the object and-homogenious multidimensional array 
print(len(array)) ## length total no of elements in the array
print(array.ndim) ## dimension of the array 1
print(array.shape)  ##dimension of array -tuples of the integer indicating the size of each dimension. 6r on c in vector 
```

```python
array2=array.reshape(3,2) ## reshape of the array 3r - 2c
print(array2)
array2.shape
```

```python
array3 = array.reshape(3,-1) ## 1Row - 1Column
print(array3)
print(array3.ndim)
```

#### **Initializing numpy array from nested Python lists**

```python
my_list2=[1,2,3,4,5]
my_list3=[2,3,4,5,6]
my_list4=[9,7,6,5,9]
mul_arr = np.array((my_list2,my_list3,my_list4))
print(mul_arr)
print(mul_arr.shape) # 3 rows and  column 
```

```python
mul_arr.reshape(1,15)
```

#### **NumPy Arthmetic Operations** 

```python
x=np.array([[1,2],[3,4]], dtype = np.float64)
y=np.array([[5,6],[7,8]], dtype=np.float64)
print(x)
print(y)
```

### **numpy.add() - performs elementwise addition between two arrays**
##### **Syntax: numpy.add(array_1,array_2)**

```python
print(x+y)
print(np.add(x,y))
```

### **numpy.subtract() - performs elementwise subtraction between two arrays**
**- Syntax : numpy.subtract(array_1,array_2)**

```python
print(x-y)
print(np.subtract(x,y))
```

### **numpy.multiply() - performs elementwise multiplication between two arrays**
**- Syntax: numpy.multiply(array_1, array_2)**

```python
print(x*y)
print(np.multiply(x,y))
```

```python
print(x.dot(y))
print(np.dot(x,y))
```

```python
print(x/y)
print(np.divide(x,y))
```

#### **numpy.sum() - returns sum of a all array elements or sum of all array elements over a given axis** 
**- Syntax: np.sum(array,axis)**

```python
print(np.sum(x)) # Compute overall sum (axis =None)
print(x)
```

```python
print(np.sum(x, axis = 0)) # sum of each column 
```

```python
print(np.sum(x,axis=1)) # each of the row == 0-row and 1-column
```
