### Set operations 
### SetSample
- A set is an unorderd collection of items
    - Every element is unique (no duplicates)
    - Sets can be used to perfrom mathematical set operations like 
    ##### (union, intersection, symmetric difference etc..)

```python
A = {'example', 24,45.6,'data',24,'data'} # set of mixed datatypes 
print(A)
```

```python
B = {24,45.6}   # set of integers
print(B)
```

```python
print(A|B)  # union of A and B is a set of all elements from both sets
A.union(B)  # using union() on B
```

```python
print(A & B)        # intersection of A and B is a set of elements that are common in both sets
A.intersection(B)  # using intersection() on B
```
